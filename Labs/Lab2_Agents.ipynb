1.
The environment is a 2D maze where # represent wall, . represent free space where the robot can stand.
Tiles S and G mark the starting and goal positions of the robot.

---

2.
Goal of the agent is to reach G, starting from S.

---

3.
Robot can more left, right, up or down, if a free space '.' is available on the destination tile.

---

4.
- Current coordinates
- Goal location and how close it is (for heuristic search)
- Location of walls (legal actions for the robot at current coordinate)
- History of actions taken.

---

5.
A simple reflex agent can easily be tricked in a setting like this. (Eg, if encounter wall, turn left can make the robot go in spirals if the maze is designed such.)
In contrast, this vehicle is a goal-based agent because its decision-making relies entirely on an explicit destination (G). It evaluates combinations of future actions through planning or pathfinding to see if they move it closer to that target objective, allowing it to navigate complex dead-ends where a simple reflex agent would get permanently stuck.


```python
# TASK 3

import heapq

def solve_maze(maze_str: str) -> str:
    # 1. Parse the string into a 2D list of characters
    grid = [
        list(line.strip())
        for line in maze_str.strip().splitlines()
        if line.strip() and not line.isspace()
    ]

    if not grid:
        return ""

    rows = len(grid)
    cols = len(grid[0])

    start_r, start_c = -1, -1
    goal_r, goal_c = -1, -1

    # Find Start (S) and Goal (G) positions
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == 'S':
                start_r, start_c = r, c
            elif grid[r][c] == 'G':
                goal_r, goal_c = r, c

    # Heuristic function (Manhattan distance)
    def manhattan(r: int, c: int) -> int:
        return abs(r - goal_r) + abs(c - goal_c)

    # 2. Setup Priority Queue (Min-Heap) and State Tracking
    # Heap stores tuples: (f_score, g_score, row, col)
    pq = []

    INF = float('inf')
    f_scores = [[INF] * cols for _ in range(rows)]
    came_from = [[None] * cols for _ in range(rows)]

    start_f = manhattan(start_r, start_c)
    f_scores[start_r][start_c] = start_f

    # Push initial state onto the heap
    heapq.heappush(pq, (start_f, 0, start_r, start_c))

    # Direction arrays for 4-way movement (Up, Down, Left, Right)
    dr = [-1, 1, 0, 0]
    dc = [0, 0, -1, 1]

    # 3. Iterate using A* Search
    while pq:
        f, g, r, c = heapq.heappop(pq)

        # If goal reached, reconstruct the path
        if r == goal_r and c == goal_c:
            curr_r, curr_c = came_from[goal_r][goal_c]
            while (curr_r, curr_c) != (start_r, start_c):
                grid[curr_r][curr_c] = '*'
                curr_r, curr_c = came_from[curr_r][curr_c]

            # Combine back into a single string
            return "\n".join("".join(row) for row in grid) + "\n"

        # Optimization: Skip stale queue entries using f_scores
        if f > f_scores[r][c]:
            continue

        # Explore neighbors
        for i in range(4):
            nr = r + dr[i]
            nc = c + dc[i]

            # Boundary check and obstacle avoidance
            if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] != '#':
                tentative_g = g + 1
                f_score = tentative_g + manhattan(nr, nc)

                # Relaxation step based on overall estimated cost (f_score)
                if f_score < f_scores[nr][nc]:
                    f_scores[nr][nc] = f_score
                    came_from[nr][nc] = (r, c)
                    heapq.heappush(pq, (f_score, tentative_g, nr, nc))

    return "No valid path found."

if __name__ == "__main__":
    maze = """
#####################
#S....#............G#
#.##....##########..#
#....##.............#
#.######.###.#.###..#
#........#..........#
#####################
    """

    print(solve_maze(maze))
```
